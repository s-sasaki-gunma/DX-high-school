## サンプルコード
- 実行時に「UserWarning: pkg_resources is deprecated as an API.」と警告が出ますが問題ありません。

```python
import pygame
import time
import os

# JupyterLabからのリモート接続用の設定
if "XDG_RUNTIME_DIR" not in os.environ:
    os.environ["XDG_RUNTIME_DIR"] = "/run/user/1000"

# 1. Pygameの初期化
pygame.init()

# 2. MP3ファイルの読み込み (※同階層に 'sample.mp3' を用意してください)
pygame.mixer.music.load("sample.mp3")

# 3. 音量の設定 (0.0 から 1.0 の間)
pygame.mixer.music.set_volume(1.0)

# 4. 再生開始 (-1 を指定するとループ再生)
pygame.mixer.music.play(-1)

# 5. 再生から n秒後 に停止
n = 12
time.sleep(n)
pygame.mixer.music.stop()
```
